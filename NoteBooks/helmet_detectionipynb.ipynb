```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
```

```python
model = YOLO("yolo11n.pt")
```

```python
results = model.train(
    data="/content/data.yaml",
    epochs=100,
    imgsz=640,
    batch=8,
    patience=20,
    project="/content/helmet_detection",
    name="experiment_1",
    plots=True
)
```

```python
fine_tuned_model = YOLO("/content/full_model.pt")
```

```python
img = cv2.imread("/content/1859298-helmet.webp")
```

```python
result = fine_tuned_model(img,save = True , conf = 0.6)
```

```python
best_model = YOLO("/content/full_model.pt")

metrics = best_model.val(
    data="/content/data.yaml",
    split="test"
)

# Display evaluation metrics
print("Precision:", metrics.box.mp)
print("Recall:", metrics.box.mr)
print("mAP50:", metrics.box.map50)
print("mAP50-95:", metrics.box.map)
```

```python
results = best_model.predict(
    source="/content/video_output.mp4",
    conf=0.25,
    imgsz=640,
    save=True,
    show=False
)
```
